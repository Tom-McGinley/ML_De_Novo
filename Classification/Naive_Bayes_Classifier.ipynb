## Bayes' Theorem
P(y|X) = [P(X|y) * P(y)]/P(X)

Where:
- P(y|X) : Posterior probability, probability of class y given features X
- P(X|y) : Likelihood, probability of features X given class y
- P(y) : Prior probability of class y
- P(X) : Marginal likelihood or evidence (of deatures X)

## Naive Assumption
Assumption: All features are conditionally independent of each other, given the class label y.

P(x1,x2,...,xn | y) = P(x1|y) * P(x2|y) * ... * P(xn|y)

Applied to Bayes Theorem:

P(y|X) = ([P(x1|y) * P(x2|y) * ... * P(xn|y)] * P(y))/P(X)


- As P(X) remains constant for each input, P(X) = 1

P(y|X) = ([P(x1|y) * P(x2|y) * ... * P(xn|y)] * P(y))

## Naive Bayes Classifier
Classifier chooses the class with the highest probability from computed posteriors for each class y:

Predicted y denoted by y^

y^ = arg_max(P(y|X))
y^ = arg_max(([P(x1|y) * P(x2|y) * ... * P(xn|y)] * P(y)))
